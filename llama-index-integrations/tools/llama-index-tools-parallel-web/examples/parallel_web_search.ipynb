# Parallel Web Search Tool Examples

This notebook demonstrates how to use the Parallel Web Search API integration with LlamaIndex tools. The Parallel API streamlines the traditional search → scrape → extract pipeline into a single, low-latency API.

## Features

- **Objective-based search**: Use natural language to describe what you're looking for
- **Query-based search**: Use specific search queries
- **Combined search**: Mix objectives with specific queries
- **Flexible processors**: Choose between "base" and "pro" processors
- **Configurable results**: Control number of results and content length

## Setup and Configuration

Import the necessary modules and set up your API key. You can get an API key from [Parallel.ai](https://parallel.ai).


```python
import os
from llama_index.tools.parallel_web import ParallelWebSearchToolSpec

# Set your Parallel API key
# Option 1: Set directly (not recommended for production)
api_key = "your_parallel_api_key_here"

# Option 2: Use environment variable (recommended)
# api_key = os.getenv("PARALLEL_API_KEY")

# Check if API key is set
if not api_key or api_key == "your_parallel_api_key_here":
    print("Please set your Parallel API key!")
    print("You can:")
    print("1. Replace 'your_parallel_api_key_here' with your actual API key")
    print("2. Set the PARALLEL_API_KEY environment variable")
    print("3. Get an API key from https://parallel.ai")
else:
    print("API key is set!")

# Initialize the tool
tool_spec = ParallelWebSearchToolSpec(api_key=api_key)
```

## Example 1: Objective-based Search

Use natural language to describe what you're looking for. This is perfect when you want to ask a question or describe your search intent in plain English.


```python
# Example 1: Objective-based search
print("Example 1: Objective-based search")
print("=" * 50)

results = tool_spec.parallel_search(
    objective="When was the United Nations established? Prefer UN's websites.",
    max_results=3,
    processor="base"
)

print(f"\nFound {len(results)} results:\n")

for i, doc in enumerate(results, 1):
    print(f"Result {i}:")
    print(f"Title: {doc.extra_info.get('title', 'N/A')}")
    print(f"URL: {doc.extra_info.get('url', 'N/A')}")
    print(f"Content: {doc.text[:200]}...")
    print("-" * 80)
```

## Example 2: Query-based Search

Use specific search queries when you know exactly what terms to search for. You can provide multiple queries to get diverse results.


```python
# Example 2: Query-based search
print("Example 2: Query-based search")
print("=" * 50)

results = tool_spec.parallel_search(
    search_queries=["SpaceX latest mission", "NASA recent discoveries"],
    max_results=5,
    max_chars_per_result=1000,
    processor="base"
)

print(f"\nFound {len(results)} results:\n")

for i, doc in enumerate(results, 1):
    print(f"Result {i}:")
    print(f"Title: {doc.extra_info.get('title', 'N/A')}")
    print(f"URL: {doc.extra_info.get('url', 'N/A')}")
    print(f"Content: {doc.text[:150]}...")
    print("-" * 80)
```

## Example 3: Combined Objective and Queries

Combine the power of both approaches: use a natural language objective along with specific search queries to get comprehensive results.


```python
# Example 3: Combined objective and queries
print("Example 3: Combined objective and queries")
print("=" * 50)

results = tool_spec.parallel_search(
    objective="Latest developments in artificial intelligence and machine learning",
    search_queries=["AI breakthroughs 2024", "machine learning advances"],
    max_results=4,
    processor="base"
)

print(f"\nFound {len(results)} results:\n")

for i, doc in enumerate(results, 1):
    print(f"Result {i}:")
    print(f"Title: {doc.extra_info.get('title', 'N/A')}")
    print(f"URL: {doc.extra_info.get('url', 'N/A')}")
    print(f"Content: {doc.text[:150]}...")
    print("-" * 80)
```

## Advanced Features and Configuration

### Using the Pro Processor

The "pro" processor provides enhanced extraction capabilities for more complex content.


```python
# Example with Pro processor
print("Advanced Example: Using Pro processor")
print("=" * 50)

results = tool_spec.parallel_search(
    objective="Compare different programming languages for web development",
    max_results=3,
    max_chars_per_result=1500,
    processor="pro"  # Using the pro processor for enhanced extraction
)

print(f"\nFound {len(results)} results with enhanced extraction:\n")

for i, doc in enumerate(results, 1):
    print(f"Result {i}:")
    print(f"Title: {doc.extra_info.get('title', 'N/A')}")
    print(f"URL: {doc.extra_info.get('url', 'N/A')}")
    print(f"Content length: {len(doc.text)} characters")
    print(f"Content preview: {doc.text[:200]}...")
    print("-" * 80)
```

## Working with Search Results

The search results are returned as LlamaIndex `Document` objects, which can be easily used in LlamaIndex workflows like building indexes, creating embeddings, or feeding into LLMs.


```python
# Working with Document objects
print("Working with Document objects")
print("=" * 50)

# Get some search results
results = tool_spec.parallel_search(
    objective="Python programming best practices",
    max_results=2,
    processor="base"
)

# Analyze the document structure
for i, doc in enumerate(results, 1):
    print(f"\nDocument {i} Analysis:")
    print(f"Type: {type(doc)}")
    print(f"Text length: {len(doc.text)} characters")
    print(f"Available metadata keys: {list(doc.extra_info.keys())}")
    
    # Extract metadata
    title = doc.extra_info.get('title', 'No title')
    url = doc.extra_info.get('url', 'No URL')
    
    print(f"Title: {title}")
    print(f"URL: {url}")
    print(f"Content sample: {doc.text[:100]}...")

# Example: Create a simple index from the search results
print(f"\n\nExample: Creating a document collection")
print("=" * 50)
all_docs = results
print(f"Created collection with {len(all_docs)} documents")
print("These documents can now be:")
print("- Added to a VectorStoreIndex")
print("- Used for RAG (Retrieval Augmented Generation)")
print("- Fed into LLMs for analysis")
print("- Processed for embeddings")
```

## Configuration Reference

Here are all the available parameters for the `parallel_search` method:

### Parameters

- **`objective`** (str, optional): Natural language description of what you're looking for
- **`search_queries`** (List[str], optional): List of specific search queries (max 5 queries, 200 chars each)
- **`max_results`** (int, optional): Maximum number of results to return (default: 10)
- **`max_chars_per_result`** (int, optional): Maximum characters per result excerpt (default: 500)
- **`processor`** (str, optional): Processor type - "base" or "pro" (default: "base")

### Notes

- At least one of `objective` or `search_queries` must be provided
- You can combine both `objective` and `search_queries` for comprehensive results
- The "pro" processor provides enhanced extraction but may have higher latency
- Results are returned as LlamaIndex `Document` objects with metadata


```python
# Example with all parameters
print("Complete Example: Using all available parameters")
print("=" * 60)

results = tool_spec.parallel_search(
    objective="Find comprehensive information about renewable energy technologies",
    search_queries=["solar power efficiency", "wind energy 2024"],
    max_results=3,
    max_chars_per_result=800,
    processor="base"
)

print(f"\nSearch completed successfully!")
print(f"Parameters used:")
print(f"- Objective: ✓")
print(f"- Search queries: ✓ (2 queries)")
print(f"- Max results: 3")
print(f"- Max chars per result: 800")
print(f"- Processor: base")
print(f"\nResults found: {len(results)}")

for i, doc in enumerate(results, 1):
    print(f"\n--- Result {i} ---")
    print(f"Title: {doc.extra_info.get('title', 'N/A')}")
    print(f"URL: {doc.extra_info.get('url', 'N/A')}")
    print(f"Content length: {len(doc.text)} characters")
    print(f"Preview: {doc.text[:120]}...")

print("\n🎉 All examples completed! You're ready to use the Parallel Web Search tool.")
```
