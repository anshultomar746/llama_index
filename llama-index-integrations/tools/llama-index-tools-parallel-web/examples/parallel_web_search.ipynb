```python
import os
from llama_index.tools.parallel_web import ParallelWebSearchToolSpec

# Set your Parallel API key
# You can get an API key from https://parallel.ai
parallel_api_key = "your_parallel_api_key_here"
# Alternatively, set the PARALLEL_API_KEY environment variable
# parallel_api_key = os.getenv("PARALLEL_API_KEY")

# Initialize the tool
tool_spec = ParallelWebSearchToolSpec(api_key=parallel_api_key)
```

```python
# Search using natural language objective
results = tool_spec.parallel_search(
    objective="When was the United Nations established? Prefer official UN websites.",
    max_results=3,
    processor="base"
)

print(f"Found {len(results)} results:\n")
for i, doc in enumerate(results, 1):
    print(f"Result {i}:")
    print(f"Title: {doc.extra_info.get('title', 'N/A')}")
    print(f"URL: {doc.extra_info.get('url', 'N/A')}")
    print(f"Content preview: {doc.text[:200]}...")
    print("-" * 80)
```
