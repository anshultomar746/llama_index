# Parallel Chat API - LlamaIndex Integration Examples

This notebook demonstrates how to use the [Parallel Chat API](https://docs.parallel.ai/resources/chat-api) with LlamaIndex for real-time web research and AI applications.

## Overview

The Parallel Chat API is a low latency web research API that brings real-time web data to interactive AI applications. Key features:

- **Real-time web research**: Get current information from the web
- **OpenAI compatibility**: Drop-in replacement with familiar interface
- **Low latency**: 3 second p50 time-to-first-token with streaming
- **JSON schema support**: Structured outputs for applications
- **Interactive optimized**: Perfect for chat interfaces and tools

**Note**: The Chat API is currently in beta with a 30 requests/minute rate limit.

### API Key Configuration

Get your API key from [Parallel.ai](https://parallel.ai) and configure it:


```python
import os
from llama_index.llms.parallel_web import ParallelWeb
from llama_index.core.llms import ChatMessage

# Set your Parallel API key
# Option 1: Set directly (not recommended for production)
api_key = "your_parallel_api_key_here"

# Option 2: Use environment variable (recommended)
# api_key = os.getenv("PARALLEL_API_KEY")

# Check if API key is set
if not api_key or api_key == "your_parallel_api_key_here":
    print("⚠️  Please set your Parallel API key!")
    print("You can:")
    print("1. Replace 'your_parallel_api_key_here' with your actual API key")
    print("2. Set the PARALLEL_API_KEY environment variable")
    print("3. Get an API key from https://parallel.ai")
else:
    print("✅ API key is set!")

# Initialize the LLM
llm = ParallelWeb(
    api_key=api_key,
    model="speed",  # The Chat API model name
    max_tokens=500
)

print("ParallelWeb LLM initialized successfully!")
```

## Example 1: Basic Web Research Chat

The main advantage of Parallel's Chat API is real-time web research. Ask questions that require current information:


```python
# Example 1: Real-time web research
messages = [
    ChatMessage(
        role="user",
        content="What are the latest developments in artificial intelligence this week?"
    )
]

response = llm.chat(messages)
print("Response:")
print(response.message.content)
```

## Example 2: JSON Response Format

A key feature of the Parallel Chat API is structured JSON responses. This is perfect for applications that need formatted data:


```python
# Define a JSON schema for structured responses
response_format = {
    "type": "json_schema",
    "json_schema": {
        "name": "research_response",
        "schema": {
            "type": "object",
            "properties": {
                "reasoning": {
                    "type": "string",
                    "description": "Step by step reasoning for the answer"
                },
                "answer": {
                    "type": "string",
                    "description": "The direct answer to the question"
                },
                "citations": {
                    "type": "array",
                    "items": {"type": "string"},
                    "description": "Sources cited to support the answer"
                }
            }
        }
    }
}

# Create LLM with JSON response format
llm_json = ParallelWeb(
    api_key=api_key,
    model="speed",
    response_format=response_format
)

# Ask a question that will return structured JSON
messages = [ChatMessage(
    role="user", 
    content="What does Parallel Web Systems do? Provide reasoning and citations."
)]

response = llm_json.chat(messages)
print("JSON Response:")
print(response.message.content)
```

## Example 3: Streaming Responses

For interactive applications, you can stream responses as they're being generated for better user experience:


```python
# Example 3: Streaming responses
messages = [
    ChatMessage(
        role="user",
        content="What are the current trends in renewable energy technology?"
    )
]

print("Question: What are the current trends in renewable energy technology?")
print("Streaming response: ", end="", flush=True)

response_stream = llm.stream_chat(messages)
for chunk in response_stream:
    print(chunk.delta, end="", flush=True)
print("\n")  # New line after streaming
```

## Example 4: Simple Completions

For straightforward prompts, you can use the `complete` method:


```python
# Example 4: Simple completion
prompt = "Explain the latest breakthroughs in quantum computing."
response = llm.complete(prompt)

print(f"Prompt: {prompt}")
print(f"Response: {response.text}")
```

## Example 5: Real-time Web Research Examples

The Chat API excels at questions requiring current information. Here are some examples:


```python
# Example 5: Current events and real-time research
current_events_prompts = [
    "What are today's top technology news stories?",
    "What are the latest updates from SpaceX missions?",
    "What are recent developments in climate change research?",
    "What is the current status of the stock market?"
]

print("Real-time Web Research Examples:")
print("=" * 50)

for prompt in current_events_prompts:
    print(f"\n🔍 Query: {prompt}")
    response = llm.complete(prompt)
    print(f"📄 Answer: {response.text[:200]}...")
    print("-" * 40)
```

## Example 6: Multi-turn Conversation

Handle multi-turn conversations with context preservation:


```python
# Example 6: Multi-turn conversation
conversation = [
    ChatMessage(
        role="user", 
        content="What is the current state of electric vehicle adoption globally?"
    ),
    ChatMessage(
        role="assistant", 
        content="Based on recent data, electric vehicle adoption is accelerating globally, with sales reaching record highs in 2024..."
    ),
    ChatMessage(
        role="user", 
        content="Which companies are leading in this space and what are their latest innovations?"
    )
]

print("Multi-turn Conversation Example:")
print("=" * 40)
print("User: What is the current state of electric vehicle adoption globally?")
print("Assistant: [Previous response context...]")
print("User: Which companies are leading in this space and what are their latest innovations?")
print()

response = llm.chat(conversation)
print("Assistant's Follow-up Response:")
print(response.message.content)
```

## Configuration Reference

### Available Parameters

When initializing `ParallelWeb`, you can configure:

- **`api_key`** (str): Your Parallel API key
- **`model`** (str): Model name - use "speed" for the Chat API
- **`max_tokens`** (int): Maximum tokens in response
- **`temperature`** (float): Sampling temperature (0.0 to 1.0)
- **`response_format`** (dict): JSON schema for structured responses
- **`api_base`** (str): API endpoint (default: "https://beta.parallel.ai")

### Rate Limits (Beta)

- **Limit**: 30 requests per minute
- **Performance**: 3 second p50 TTFT with streaming
- **Use Cases**: Interactive applications, chat interfaces

### When to Use Parallel Chat API

✅ **Perfect for:**
- Real-time web research applications
